```python
# ================================      VI.0)chargement des recettes ayant pas de valeur vide(recipes_clean.csv)
try:
    recipes_clean = pd.read_csv("recipes_clean.csv")
    print("Chargement des donnees de recettes clean  reussis !")
except Exception as e:
    print("Une erreur est survenue l'or de l'operation: ",e)
```

```python

# ========================       VI.1) preparons les colonnes ingrdients et tags a algorithme de recherche (TF-IDF)      ================
import ast
# ====================================     VI.1.a) Mettons les ingredients au tyoe str      ==============================================
recipes_clean['ingredients'] = recipes_clean['ingredients'].apply(ast.literal_eval)
recipes_clean['ingredients'] = recipes_clean['ingredients'].apply(lambda x: " ".join(x))

# =============================================      VI.1.b) Mettons les tagsau type str  ==================================================
recipes_clean['tags'] = recipes_clean['tags'].apply(ast.literal_eval)
recipes_clean['tags'] = recipes_clean['tags'].apply(lambda x: " ".join(x))

# ===============================       VI.1.c) cree une colonnes de combinations pour les tags, les ingredients         ================
recipes_clean['recombined_text'] = (recipes_clean['ingredients'] + " " + recipes_clean['tags'])

# ============================  VI.2)mise a jour de recipes avec les Attributs selectionne pour le systeme de recommandation    =============
recipes_clean = recipes_clean[['name','minutes', 'tags','ingredients','recombined_text']]

# =============================================      Sauvegarder les modification dans le fichier   ==================================
try:
    recipes_clean.to_csv('recipes_clean.csv', index = False)
    print("sauvegarde reussis dans le fichier recipes_clean.csv")
except Exception as e:
    print("Une erreur est survenue l'or de l'oprations :", e)
```

```python
# =======================================================================================================================================
#                                               IV.Systeme de recommandation     
# =======================================================================================================================================


# ======================================     IV.1.Recommandation par ingredients    ====================================================

# ========================================     IV.1.a) TF-IDF(Term Frequency-Inverse Document Frequency     ============================
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(recipes_clean['recombined_text'])
```

```python
# ==================================     Fonction de recommandation filtrer par temps de preparation      ==========

def recommend_recipes(ingredients_input, max_time=60, tag=None, top_n=5):
    
    input_text = " ".join(ingredients_input)
    
    input_vec = vectorizer.transform([input_text])
    
    similarity = cosine_similarity(input_vec, tfidf_matrix)[0]
    
    # copier dataframe
    filtered = recipes_clean.copy()
    filtered['score'] = similarity
    
    # filtrage temps
    filtered = filtered[filtered['minutes'] <= max_time]
    
    # filtrage tag
    if tag:
        filtered = filtered[
            filtered['tags'].str.contains(tag, case=False, na=False)
        ]
    
    results = filtered.sort_values(by='score', ascending=False)
    
    return results[['name', 'ingredients', 'minutes', 'score']].head(top_n)
    
# =============================     Test de la fonction de recommandation    ========================================================
reco = recommend_recipes(
    ingredients_input=["chicken", "garlic"],
    max_time=45,
    tag="weeknight",
    top_n=5
)
print("| ----------------------------------------------------------------------------------------|")
print("| Nombre de résultats :", len(reco))
print("| ----------------------------------------------------------------------------------------|")
print(reco)
print("| ----------------------------------------------------------------------------------------|")
```
